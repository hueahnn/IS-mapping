```python
import pandas as pd
import zipfile

def load_sample(genome_acc, base_dir="/n/scratch/users/h/hua575/atb_filtered/gene_disruption_v2"):
    with zipfile.ZipFile(f"{base_dir}/{genome_acc}.zip") as z:
        dfs = [pd.read_csv(z.open(name), sep="\t") for name in z.namelist()]
    return pd.concat(dfs, ignore_index=True)

sample_df = load_sample("SRR15366008")
```

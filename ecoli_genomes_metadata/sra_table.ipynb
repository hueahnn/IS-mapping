```python
import polars as pl

# read in biosample IDs
with open("ecoli_biosampleids.txt") as f:
    biosample_ids = [line.strip() for line in f if line.strip() and line.strip() != "null"]

print(f"Total BioSample IDs: {len(biosample_ids)}")

# read in SRA table from NCBI
sra_table = pl.scan_csv(
    "/n/scratch/users/h/hua575/ecoli_sra_filtered.tab",
    separator="\t"
)

# Only load columns you need - saves memory on a 31GB file
sra_table = sra_table.select(["Accession", "Submission", "Bases", "Sample", "Type", "BioSample"])

# Filter to RUN rows only before collecting into memory
sra_table = sra_table.filter(
    (pl.col("Bases") != "-") &
    (pl.col("BioSample") != "-") &
    (pl.col("BioSample").is_in(biosample_ids))
).collect()
```

```python
# checking ISfinder DB from last summer and this summer, what's different bc 2026 DB is smaller
from Bio import SeqIO

seqs1 = {r.id: str(r.seq) for r in SeqIO.parse("../ISfinder_database-master/IS.database.fa", "fasta")}
seqs2 = {r.id: str(r.seq) for r in SeqIO.parse("../ISfinder_database-master_2026/IS.database.fa", "fasta")}

shared_ids    = seqs1.keys() & seqs2.keys()
only_in_1     = seqs1.keys() - seqs2.keys()
only_in_2     = seqs2.keys() - seqs1.keys()

print(f"Shared IDs:       {len(shared_ids)}")
print(f"Only in file1:    {len(only_in_1)}")
print(f"Only in file2:    {len(only_in_2)}")

same_seq = {i for i in shared_ids if seqs1[i] == seqs2[i]}
diff_seq = {i for i in shared_ids if seqs1[i] != seqs2[i]}
print(f"Same ID, same seq:  {len(same_seq)}")
print(f"Same ID, diff seq:  {len(diff_seq)}")
```

```python
# checking which seqs are duplicates
from Bio import SeqIO
from collections import Counter

def parse_fasta(path):
	records = list(SeqIO.parse(path, "fasta"))
	ids  = [r.id for r in records]
	seqs = [str(r.seq) for r in records]
	return ids, seqs

ids1, seqs1 = parse_fasta("../ISfinder_database-master_2026/IS.database.fa")

# duplicate seqs within each file
seq_counts1 = Counter(seqs1)

dup_seqs1 = {seq: cnt for seq, cnt in seq_counts1.items() if cnt > 1}

print(f"file1 duplicate seqs: {len(dup_seqs1)} duplicates")				
```
